```python
# function 1 
def english_predicate(sentence)  :
    sentence=sentence.lower().strip(" .")
    all_words = ("all", "every", "each")
    some_words = ("some", "a", "an")
    none_words = ("no", "none")

    
    for all in all_words :
        if(sentence.startswith(f"{all} ")):
            sentence=sentence.replace(f"{all}","")
            if "are" in sentence :
                sentence=sentence.split("are")
            elif "is" in sentence :
                sentence=sentence.split("is") 
            predicate=print(f"∀x({sentence[0]}(x)->{sentence[1]}(x))")
            return predicate 
    for some in some_words : 
        if sentence.startswith(f"{some} "):
            sentence = sentence.replace(f"{some}","")
            if "are" in sentence :
                sentence=sentence.split("are")
            elif "is" in sentence :
                sentence=sentence.split("is") 
            predicate=print(f"∃x({sentence[0]}(x) ∧ {sentence[1]}(x))")
            return predicate
    for none in none_words:
        if sentence.startswith(f"{none} "):
            sentence = sentence.replace(f"{none} ","")
            if "are" in sentence :
                sentence=sentence.split("are")
            elif "is" in sentence :
                sentence=sentence.split("is") 
            predicate=print(f"¬∃x ({sentence[0]}(x) ∧ {sentence[1]}(x))")
            return predicate

        
```

```python
# Test English to Predicate
print(english_predicate("Every human is mortal."))
print(english_predicate("Each student is tired."))
print(english_predicate("A dog is friendly."))
print(english_predicate("No birds are mammals."))
print(english_predicate("None cats are reptiles."))


```

```python
def predecate_englsih(predicate):
    if predicate.startswith("∀x") : 
        predicate=predicate[4:-1]
        if("->" in predicate) : 
            predicate=predicate.split("->")
            predicate[0]=predicate[0].replace("(x)", "")
            predicate[1]=predicate[1].replace("(x)","")
            english_sentence=print(f"All {predicate[0]}s are {predicate[1]}s.")
            return english_sentence
    elif predicate.startswith("∃x"):
        predicate = predicate[4:-1]
        if "∧" in predicate:
            predicate = predicate.split(" ∧ ")
            predicate[0] = predicate[0].replace("(x)", "")
            predicate[1] = predicate[1].replace("(x)", "")
            english_sentence= print(f"Some {predicate[0]}s are {predicate[1]}s.")
            return english_sentence
    elif predicate.startswith("¬∃x"):
        predicate = predicate[4:-1]
        if " ∧ " in predicate :
            predicate = predicate.split(" ∧ ")
            predicate[0] = predicate[0].replace("(x)", "")
            predicate[1] = predicate[1].replace("(x)", "")
            return f"No {predicate[0]}s are {predicate[1]}s."   
```

```python
#test the predicate english 
print(predecate_englsih("∀x (human(x) -> mortal(x))"))
print(predecate_englsih("∃x (dog(x) ∧ friendly(x))"))
print(predecate_englsih("¬∃x (cat(x) ∧ reptile(x))"))
```
